# Stock Price Prediction — LSTM training notebook

Trains the LSTM model used by the Flask app (`stock_dl_model.h5`). This version fixes two
correctness bugs found in the original notebook, called out inline below:

1. **Scaler leakage** — the original re-fit `MinMaxScaler` on the test window
   (`scaler.fit_transform(final_df)`), which both leaks test-set statistics into the transform
   and produces a different scale than the one the model trained on. Fixed to `scaler.transform(...)`,
   reusing the scaler fit only on `data_training`.
2. **Hardcoded de-scaling factor** — the original inverted the scaling with a magic constant
   (`scaler_factor = 1 / 0.0035166`) computed for one specific dataset, which silently produces
   wrong prices for any other ticker or date range. Fixed to `scaler.inverse_transform(...)`.

Also added: `yfinance` MultiIndex column flattening (newer `yfinance` versions return
`(field, ticker)` column tuples even for a single ticker), plus accuracy metrics and a
forward-looking forecast section so the notebook mirrors what `app.py` does at inference time.

## 4. Train / test split and scaling

```python
# Training & Testing
data_training = pd.DataFrame(df['Close'][0:int(len(df) * 0.70)])
data_testing = pd.DataFrame(df['Close'][int(len(df) * 0.70): int(len(df))])
```

```python
data_training.shape
```

```python
data_testing.shape
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0, 1))
```

```python
# Fit the scaler ONCE, on training data only. Everything downstream — the test
# window and any future forecast — must only ever call scaler.transform(...),
# never scaler.fit_transform(...) again, or test-set statistics leak into the
# scale and the model effectively gets evaluated on a different distribution
# than it trained on.
data_training_array = scaler.fit_transform(data_training)
```

```python
data_training_array
```

```python
data_training_array.shape[0]
```

```python
x_train = []
y_train = []

for i in range(100, data_training_array.shape[0]):
    x_train.append(data_training_array[i - 100:i])
    y_train.append(data_training_array[i, 0])

x_train, y_train = np.array(x_train), np.array(y_train)
```

```python
x_train.shape
```

## 5. Model

```python
# Model Building
from keras.layers import Dense, Dropout, LSTM
from keras.models import Sequential
```

LSTM Input -> 3D Array (batch_size, time_steps, seq_len)
LSTM 2D OR 3D
2D -> (batch_size, units)
3D -> (batch_size, time_steps, units)

```python
model = Sequential()

model.add(LSTM(units=50, activation='relu', return_sequences=True, input_shape=(x_train.shape[1], 1)))
model.add(Dropout(0.2))

model.add(LSTM(units=60, activation='relu', return_sequences=True))
model.add(Dropout(0.3))

model.add(LSTM(units=80, activation='relu', return_sequences=True))
model.add(Dropout(0.4))

model.add(LSTM(units=120, activation='relu'))
model.add(Dropout(0.5))

model.add(Dense(units=1))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(x_train, y_train, epochs=50)
```

## 6. Backtest on the held-out test window

```python
past_100_days = data_training.tail(100)
```

```python
final_df = pd.concat([past_100_days, data_testing], ignore_index=True)
```

```python
final_df.head()
```

```python
# FIX: transform only - reuse the scaler fit on training data above.
# The original notebook called scaler.fit_transform(final_df) here, which
# re-fits on the test window (data leakage) and silently changes the scale
# the model's outputs need to be inverted with.
input_data = scaler.transform(final_df)
```

```python
x_test = []
y_test = []

for i in range(100, input_data.shape[0]):
    x_test.append(input_data[i - 100:i])
    y_test.append(input_data[i, 0])

x_test, y_test = np.array(x_test), np.array(y_test)
```

```python
x_test.shape
```

```python
y_predicted = model.predict(x_test)
```

```python
y_predicted.shape
```

```python
scaler.scale_
```

```python
# FIX: use the scaler's own inverse_transform instead of a hardcoded magic
# constant (scaler_factor = 1 / 0.0035166). That constant only happens to be
# correct for the exact dataset it was computed from - a different ticker or
# date range gives a different scaler.scale_ and the hardcoded number would
# silently produce wrong prices.
y_predicted_actual = scaler.inverse_transform(y_predicted).flatten()
y_test_actual = scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(y_test_actual, label='Original Price', linewidth=1)
plt.plot(y_predicted_actual, label='Predicted Price', linewidth=1)
plt.legend()
plt.show()
```

## 7. Accuracy metrics

Same metrics `app.py` reports for a live prediction, so a run in this notebook and a run in the app are comparable.

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

rmse = float(np.sqrt(mean_squared_error(y_test_actual, y_predicted_actual)))
mae = float(mean_absolute_error(y_test_actual, y_predicted_actual))
mape = float(np.mean(np.abs((y_test_actual - y_predicted_actual) / y_test_actual)) * 100)
r2 = float(r2_score(y_test_actual, y_predicted_actual))

print(f"RMSE: {rmse:.2f}")
print(f"MAE:  {mae:.2f}")
print(f"MAPE: {mape:.2f}%")
print(f"R^2:  {r2:.3f}")
```

## 8. Forecast forward

Mirrors `forecast_future()` in `app.py`: iteratively predict the next N trading days, feeding each prediction back in as if it were an observed close.

```python
def forecast_future(close_series, scaler, model, days=30, window=100):
    last_window = close_series.tail(window).values.reshape(-1, 1)
    scaled_window = scaler.transform(last_window)

    preds_scaled = []
    window_data = scaled_window.copy()
    for _ in range(days):
        x = window_data[-window:].reshape(1, window, 1)
        next_scaled = model.predict(x, verbose=0)[0]
        preds_scaled.append(next_scaled[0])
        window_data = np.vstack([window_data, next_scaled.reshape(1, 1)])

    return scaler.inverse_transform(np.array(preds_scaled).reshape(-1, 1)).flatten()


forecast_days = 30
forecast_prices = forecast_future(df['Close'], scaler, model, days=forecast_days)
forecast_prices
```

```python
plt.figure(figsize=(12, 6))
plt.plot(range(len(y_test_actual)), y_test_actual, label='Actual (test)', linewidth=1)
plt.plot(range(len(y_predicted_actual)), y_predicted_actual, label='Predicted (test)', linewidth=1)
plt.plot(range(len(y_test_actual), len(y_test_actual) + forecast_days), forecast_prices,
         label='Forecast (future)', linewidth=1, linestyle='--')
plt.legend()
plt.show()
```
